# Array Basics + Types

```python
arr = np.arange(1, 21)

print("Array : ", arr)
print("Shape : ", arr.shape)
print("Dtype : ", arr.dtype)
print("Min value : ", arr.min())
print("Max Value : ", arr.max())
print("Sum : ", arr.sum())
print("Mean : ", arr.min())

arr = arr.astype('float64')
print("Array : ", arr)
print("Dtype : ", arr.dtype)
```

# Slicing + Boolean masking

```python
arr = np.random.randint(1, 101, size=50)

print("Random array : ", arr)

even = arr[arr % 2 == 0]
print("\nEven numbers : ", even)

div_by_3_and_gt_50 = arr[(arr % 3 == 0) & (arr > 50)]
print("\nElements greater that 50 and divisiible by 3 : ", div_by_3_and_gt_50)

replaced = np.where(arr < 20, 20, arr)
print("\nArray after replacing less that 20 values with 20 : ", replaced)
```

# Reshape + Axis Opeartions

```python
arr = np.arange(1, 61)
arr = arr.reshape(5, 12)

print("Original array : ", arr)

row_sum = arr.sum(axis=1)
print("Row wise Sum : ", row_sum)

col_mean = arr.mean(axis=0)
print("Columns wise mean : ",col_mean)

standard_deviation = arr.std()
print("Standard Deviation : ", standard_deviation)

row, col = np.where(arr == arr.max())
print("Position of max element : ", row, col)
```

# Broadcasting Practice

```python
# (4, 5) matrix of random floats
A = np.random.rand(4, 5)

# (5,) vector
v = np.random.rand(5)

# Adding vector v to every row using broadcasting
B = A + v

# Normalize each row to sum to 1
row_sums = B.sum(axis=1, keepdims=True)

# Avoiding division by zero by replacing 0 with 1
row_sums = np.where(row_sums == 0, 1, row_sums)

B_normalized = B / row_sums

print("Original matrix A:\n", A)
print("\nVector v:\n", v)
print("\nAfter broadcasting addition B:\n", B)
print("\nRow-normalized matrix:\n", B_normalized)
print("\nRow sums (should be 1):\n", B_normalized.sum(axis=1))

```

# Matrix Ops (Transpose, Identity, Dot)

```python
# Generate matrices
A = np.random.rand(3, 4)
B = np.random.rand(4, 2)

# Computing matrix product A @ B
C = A @ B

# Verifying transpose property
transpose_check = np.allclose((A.T).T, A)

# Create identity matrix and verify A @ I = A (shape permitting)
I = np.eye(4)
A_identity = A @ I

print("Matrix A (3x4):\n", A)
print("\nMatrix B (4x2):\n", B)
print("\nA @ B (3x2):\n", C)

print("\n(A.T).T equals A:", transpose_check)

print("\nIdentity matrix I (4x4):\n", I)
print("\nA @ I:\n", A_identity)
print("\nA @ I equals A:", np.allclose(A_identity, A))
```

# Sorting + Top-k without Sort

```python
arr = np.random.rand(100)

k = 10

# Get indices of top-k elements
topk_idx = np.argpartition(arr, -k)[-k:]

# Sort those top-k indices by value (descending)
top_elements = arr[topk_idx]
sorted_topk_idx = topk_idx[np.argsort(top_elements)[::-1]]

topk_values = arr[sorted_topk_idx]

print("Top 10 values:")
for i, v in zip(sorted_topk_idx, topk_values):
    print(f"Index: {i:3d}, Value: {v:.6f}")


```

# Missing Value Simulation

```python
arr = np.random.rand(40)

num_missing = int(0.2 * arr.size)  # 20% of 40 = 8
missing_indices = np.random.choice(arr.size, size=num_missing, replace=False)
arr[missing_indices] = np.nan

mean_val = np.nanmean(arr)

median_val = np.nanmedian(arr)
arr[np.isnan(arr)] = median_val

print("Indices of NaNs:", missing_indices)
print("Mean ignoring NaNs:", mean_val)
print("Median used to replace NaNs:", median_val)
print("Array after filling NaNs:\n", arr)
```

# Linear Regression From Scratch

```python
n_samples = 200

X = np.random.rand(n_samples, 1)

noise = np.random.randn(n_samples, 1) * 0.5

y = 3 * X + 5 + noise

X_design = np.hstack([
    np.ones((n_samples, 1)),
    X
])

# Normal Equation
#    theta = (X^T X)^(-1) X^T y
theta = np.linalg.inv(
    X_design.T @ X_design
) @ X_design.T @ y


intercept = theta[0, 0]
slope = theta[1, 0]

print(f"Estimated intercept: {intercept:.4f}")
print(f"Estimated slope: {slope:.4f}")
```

# Time Series Rolling Window Stats

```python
n_days = 365

sales = np.random.normal(loc=100, scale=15, size=n_days)


def rolling_window(arr, window):
    """
    Create rolling window view of a 1D array
    """
    shape = (arr.size - window + 1, window)
    strides = (arr.strides[0], arr.strides[0])
    return np.lib.stride_tricks.as_strided(arr, shape=shape, strides=strides)

roll7 = rolling_window(sales, 7)
rolling_7_mean = np.full(n_days, np.nan)
rolling_7_mean[6:] = roll7.mean(axis=1)

roll30 = rolling_window(sales, 30)

rolling_30_mean = np.full(n_days, np.nan)
rolling_30_std = np.full(n_days, np.nan)

rolling_30_mean[29:] = roll30.mean(axis=1)
rolling_30_std[29:] = roll30.std(axis=1)

anomalies = (
    sales > (rolling_30_mean + 2 * rolling_30_std)
)

anomaly_days = np.where(anomalies)[0]

print(f"Total anomaly days detected: {len(anomaly_days)}")
print("First 10 anomaly day indices:", anomaly_days[:10])
```
